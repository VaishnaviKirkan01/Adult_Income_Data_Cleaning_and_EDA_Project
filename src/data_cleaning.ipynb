```python
import pandas as pd
import numpy as np
```

```python
# giving the column names to the datset

columns=[
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]
```

```python
# adult.data dataset in the train_df

train_df=pd.read_csv(
    "C:\\Users\\vaish\\OneDrive\\Documents\\Adult_Income_Data_Cleaning_and_EDA\\data\\adult.data",
    names=columns,
    sep=",",
    skipinitialspace=True
)

train_df.head()
```

```python
print(train_df.shape)
```

```python
# adult.test dataset in the test_df

test_df=pd.read_csv(
    "C:\\Users\\vaish\\OneDrive\\Documents\\Adult_Income_Data_Cleaning_and_EDA\\data\\adult.test",
    names=columns,
    sep=",",
    skipinitialspace=True,
    skiprows=1
)

test_df.head()
```

```python
# # converting the tarin_df and test_df into train and test

# train_df["source_split"]="train"
# test_df["source_split"]="test"
# df
```

```python
# now concat both of them

df=pd.concat(
    [train_df,test_df],
    ignore_index=True
)
```

```python
print(df.shape)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.describe()
```

```python
# Removing the leading and trailing spaces from all the catogrical columns

columns_cat=df.select_dtypes(include="object").columns

for col in columns_cat:
    df[col]=df[col].str.strip()
```

```python
# contains ? in place of the unknown values

df=df.replace("?",np.nan)
```

```python
df.isnull().sum()
```

```python
missing_values=pd.DataFrame(
    {
        "Missing_Count": df.isnull().sum(),
        "Missing_Percentage":df.isnull().mean()*100
    }
)
missing_values=missing_values[missing_values["Missing_Count"]>0]

print(missing_values)
```

```python
# filling the missing values with unknown
missing_cat=[
    "workclass",
    "occupation",
    "native_country"
]

for i in missing_cat:
    df[i]=df[i].fillna("Unknown")

df[missing_cat].isnull().sum()
```

```python
# income Column

df["income"]=df["income"].replace(".","",regex=False)
```

```python
# filtering out the higher income than 50k

df["high_income"]=(df["income"]==">50K").astype(int)

df[["income", "high_income"]].head(10)
```

```python
# numerical columns

numeric_col=[
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

for i in numeric_col:
    df[i]=pd.to_numeric(df[i],errors="coerce")

df[numeric_col].dtypes
df[numeric_col].isnull().sum()
df[numeric_col].duplicated().sum()
```

```python
# creating the different age groups

df["age_group"]=pd.cut(df["age"],
                       bins=[1,25,35,45,55,65,100],
                       labels=["18-25",'26-35','36-45','46-55','56-65','66+'])

df['age_group'].value_counts()
```

```python
# creating the different working hour groups

df["hour_group"]=pd.cut(df["hours_per_week"],
                        bins=[0,20,40,60,150],
                        labels=['0-20','21-40','41-60','61+']
)


df["hour_group"].value_counts()
```

```python
# creating the net capital column from capital gain and captial loss

df["net_capital"]=(df["capital_gain"]-df["capital_loss"])

df[["capital_gain","capital_loss","net_capital"]].head()
```

```python
df.isnull().sum()

```

```python
# final inspection of the dataset

print("Shape: ",df.shape)

print("Columns: ",df.columns)

print("Data Types: ",df.dtypes)

print("Duplicated: ",df.duplicated().sum())
```
